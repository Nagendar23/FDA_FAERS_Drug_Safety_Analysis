### fix age_cod (age unit) inconsistency (some are in hours, days, months and years => turn all of them into years )

```python
def convert_age_to_years(row):
    if pd.isna(row['age']) or pd.isna(row['age_cod']):
        return np.nan
    conversions = {'DEC':10, 'YR':1, 'MON':1/12, 'WK':1/52, 'DY':1/365, 'HR':1/8760 }
    return row['age'] * conversions.get(row['age_cod'], np.nan)
demo_clean['age_years'] = demo_clean.apply(convert_age_to_years, axis=1)
```

### Sanity check on Weight (wt_cod)

```python
def convert_wt_to_kg(row):
    if pd.isna(row['wt']) or pd.isna(row['wt_cod']):
        return np.nan
    if row['wt_cod'] == 'LBS':
        return row['wt'] * 0.453592
    elif row['wt_cod'] == 'KG':
        return row['wt']
    return np.nan
demo_clean['wt_kg'] = demo_clean.apply(convert_wt_to_kg, axis=1)
```

```python
import time
import requests

def get_generic_name(drug_name):
    try:
        rxcui_resp=requests.get(f"https://rxnav.nlm.nih.gov/REST/rxcui.json?name={drug_name}")
        rxcui = rxcui_resp.json()['idGroup']['rxnormId'][0]
        related_resp = requests.get(f"https://rxnav.nlm.nih.gov/REST/rxcui/{rxcui}/related.json?tty=IN")
        ingredient = related_resp.json()['relatedGroup']['conceptGroup'][0]['conceptProperties'][0]['name']
        return ingredient
    except (KeyError, IndexError, requests.exceptions.RequestException):
        return None
```

```python
for name in ['DUPIXENT', 'ASPIRIN', 'TYLENOL']:
    print(name, get_generic_name(name))
    time.sleep(0.5)
```

```python
top_500_names = top_drugs.index.tolist()

generic_map={}
for i, name in enumerate(top_500_names):
    generic_map[name] = get_generic_name(name)
    time.sleep(0.5)
    if i%50==0:
        print(f'{i} of 500 done' )
print('finished')
```

```python
found = sum(1 for v in generic_map.values() if v is not None)
print(f'{found} out of {len(generic_map)} drug names successfully mapped ')
```

```python
failed = [k for k, v in generic_map.items() if v is None]
print(failed[:20])
```

```python
def get_final_generic(row_name):
    mapped = generic_map.get(row_name)
    return mapped if mapped else row_name  # fallback to orginal if not mappring found
drug['generic_name'] = drug['drugname_clean'].apply(
    lambda x: get_final_generic(x) if x in generic_map else x
)
print(drug['generic_name'].nunique())
```

some drugs will naturally have more death or serious outcomes bcz they are prescribed to more people or prescribed to serious cases, so if we count the drugs alone that has deaths the result can be wrong. So it is like 'does a drug have higher rate of serious outcomes than we expect, relative to its total report volume

```python
# join DRUG(primary suspect onlyy) with OUTC
ps_drugs = drug[drug['is_primary_suspect']][['primaryid','generic_name']]
outc_small = outc[['primaryid','is_death']]

drug_outcome = ps_drugs.merge(outc_small, on='primaryid', how='left')
drug_outcome['is_death'] = drug_outcome['is_death'].fillna(False).astype(bool)
print(drug_outcome.shape)
print(drug_outcome.head())
```

### Calculate the death rate per drug (only drugs with enough volume to be in meaningful)

```python
drug_stats = drug_outcome.groupby('generic_name').agg(
    total_reports=('primaryid', 'count'),
    death_reports = ('is_death','sum')
)
drug_stats['death_rate'] = drug_stats['death_reports'] / drug_stats['total_reports']
drug_stats_filtered = drug_stats[drug_stats['total_reports'] >= 500]
print(drug_stats_filtered.shape)
```

```python
# overall avg death rate and highest/lowest outliers
overall_rate = drug_outcome['is_death'].mean()
print('overall avg death rate: ', overall_rate)
print(drug_stats_filtered.sort_values('death_rate', ascending=False).head(15))
```

```python
# making a statistical test to make sure this is not random noise
from statsmodels.stats.proportion import proportions_ztest

fentanyl_deaths = drug_stats_filtered.loc['fentanyl', 'death_reports']
fentanyl_total = drug_stats_filtered.loc['fentanyl', 'total_reports']
count = [fentanyl_deaths, drug_outcome['is_death'].sum()]
nobs = [fentanyl_total, len(drug_outcome)]
stat, pval = proportions_ztest(count, nobs)
print('z-statistic: ', stat)
print('p-value: ', pval)
```

```python
drug_stats_filtered.sort_values('death_rate', ascending=False).to_csv('drug_death_rates_2025.csv')
print('saved')
```

```python
import pandas as pd
import pymysql

connection = pymysql.connect(host='localhost', user='root', password='root1234', database='faers_2025')
cursor = connection.cursor()

cursor.execute("TRUNCATE TABLE reac")
connection.commit()

chunksize = 50000
total_inserted = 0

for chunk in pd.read_csv('reac_cleaned_2025.csv', chunksize=chunksize):
    data = list(chunk[['primaryid', 'caseid', 'pt']].itertuples(index=False, name=None))
    cursor.executemany("INSERT INTO reac (primaryid, caseid, pt) VALUES (%s, %s, %s)", data)
    connection.commit()
    total_inserted += len(chunk)
    print(total_inserted, 'rows inserted so far')

connection.close()
print('Done')
```

```python
import csv

count_with_commas_in_field = 0
with open('reac_cleaned_2025.csv', encoding='utf-8') as f:
    reader = csv.reader(f)
    next(reader)  # skip header
    for row in reader:
        if ',' in row[2]:  # pt column
            count_with_commas_in_field += 1

print(count_with_commas_in_field, 'rows have a comma inside the pt field')
```
